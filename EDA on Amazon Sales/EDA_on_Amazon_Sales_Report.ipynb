## *EDA*

```python
x = sns.countplot(data = df, x = 'Size')
```

```python
x = sns.countplot(data = df, x = 'Size')
for bars in x.containers:
  x.bar_label(bars)
```

*Note : From above Graph you can see that most of the people buy M-Size*

```python
df.groupby(['Size'], as_index=False)['Qty'].sum().sort_values(by = 'Qty', ascending=False)
```

```python
df1 = df.groupby(['Size'], as_index=False)['Qty'].sum().sort_values(by = 'Qty', ascending=False)
sns.barplot(data = df1, x = 'Size', y = 'Qty')
```

*Note: From above Graph you can see that most of the Qty buys M-Size in the sales*

```python
sns.countplot(data = df, x = 'Courier Status', hue = 'Status')
```

```python
plt.figure(figsize = (10,5))
x = sns.countplot(data = df, x = 'Courier Status', hue = 'Status')
plt.show()
```

*Note: From above Graph the majority of the orders are shipped through the courier.*

```python
df['Size'].hist()
```

```python
df['Category'] = df['Category'].astype(str)
plt.figure(figsize = (10,5))
plt.hist(x = df['Category'], bins = 30, edgecolor = 'Black')
plt.xticks(rotation = 90)
plt.show()
```

*Note : From above Graph you can see that most of the buyers are T-shirt*

```python
x = df['B2B'].value_counts()
plt.pie(x, labels = x, autopct = '1.1f%%')
plt.show()
```

```python
x = df['B2B'].value_counts()
plt.pie(x, labels = x.index, autopct = '1.1f%%')
plt.show()
```

```python
a = df['Fulfilment'].value_counts()
fig, x = plt.subplots()
x.pie(a, labels = a.index, autopct='%1.1f%%', radius=0.7, wedgeprops=dict(width=0.6))
x.set(aspect = 'equal')
plt.show()
```

*Note: From above chart you can see that most of the Fulfilment are amazon*

```python
x = df['Category']
y = df['Size']

# Plot the scatter plot
plt.scatter(x, y)
plt.xlabel('Category ')
plt.ylabel('Size')
plt.title('Scatter Plot')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='ship-state')
plt.xlabel('ship-state')
plt.ylabel('count')
plt.title('Distribution of State')
plt.xticks(rotation=90)
plt.show()
```

```python
top_10_state = df['ship-state'].value_counts().head(10)
plt.figure(figsize=(12, 6))
sns.countplot(data=df[df['ship-state'].isin(top_10_state.index)], x='ship-state')
plt.xlabel('ship-state')
plt.ylabel('count')
plt.title('Distribution of  State')
plt.xticks(rotation=45)
plt.show()
```

*Note: From above Graph you can see that most of the buyers are Maharashtra state*
